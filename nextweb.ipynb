```python
import numpy as np
import pandas as pd
import time
from nltk.stem import PorterStemmer

#load dataset buku
df = pd.read_csv('fix.csv')
print(df.head())
```

```python
data = df[['title','authors','categories','description']]
print(data.head())
```

```python
data.isnull().sum()
```

```python
data = data.fillna('')
print(data.isnull().sum())
```

```python
data['combined'] = data['title'] + " " + data['authors'] + " " + data['categories'] + " " + data['description']
print(data['combined'].iloc[0])
```

```python
data['combined'] = data['combined'].str.lower()
print(data['combined'].iloc[0])
```

```python
data['tokens'] = data['combined'].apply(lambda x: x.split())
print(data['tokens'].iloc[0])
```

```python
stopwords = ['and','a','about','the','of','is','that']
data['filtered'] = data['tokens'].apply(lambda x: [w for w in x if w not in stopwords])
print(data['filtered'].iloc[0])
```

```python
stemmer = PorterStemmer()
data['stemmed'] = data['filtered'].apply(
    lambda x: [stemmer.stem(word) for word in x]
)
print(data['stemmed'].iloc[0])
```

```python
data['final'] = data['filtered'].apply(lambda x: ' '.join(x))
print(data['final'].iloc[0])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Hyperparameter tuning
tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=5000,
    ngram_range=(1,2),
    min_df=2,
    max_df=0.8
)
```

```python
# TF-IDF matrix
tfidf_matrix = tfidf.fit_transform(data['final'])
```

```python
# Cosine similarity
cosine_sim = cosine_similarity(tfidf_matrix)
```

```python
# cek hasil
print(tfidf_matrix.shape)
print(cosine_sim[:5])
```

```python
def rekomendasi(judul):
    judul = str(judul).strip().lower()
    
    data['title_clean'] = data['title'].astype(str).str.lower()
    
    hasil = data[data['title_clean'].str.contains(judul, na=False)]
    
    if hasil.empty:
        print("Judul tidak ditemukan")
        return
    
    idx = hasil.index[0]
    
    skor = list(enumerate(cosine_sim[idx]))
    skor = sorted(skor, key=lambda x: x[1], reverse=True)
    
    print("Rekomendasi untuk:", data.loc[idx, 'title'])
    print("-" * 30)
    
    count = 0
    for i in skor:
        if i[0] != idx:
            print(data.loc[i[0], 'title'], "->", round(i[1], 3))
            count += 1
        if count == 3:
            break
```

```python
rekomendasi("gilead")
```

```python
def precision_at_k(idx, cosine_sim, data, k=3):

    skor = list(enumerate(cosine_sim[idx]))
    skor = sorted(skor, key=lambda x: x[1], reverse=True)

    rekom = [i[0] for i in skor[1:k+1]]

    kategori_asli = set(
        str(data.loc[idx, 'categories']).lower().split()
    )

    relevan = 0

    for i in rekom:

        kategori_rekom = set(
            str(data.loc[i, 'categories']).lower().split()
        )

        if len(kategori_asli & kategori_rekom) > 0:
            relevan += 1

    return relevan / k
```

```python
def ndcg_at_k(idx, cosine_sim, data, k=3):
    skor = list(enumerate(cosine_sim[idx]))
    skor = sorted(skor, key=lambda x: x[1], reverse=True)

    rekom = [i[0] for i in skor[1:k+1]]

    kategori_asli = set(
        str(data.loc[idx, 'categories']).lower().split()
    )

    relevansi = []

    for i in rekom:

        kategori_rekom = set(
            str(data.loc[i, 'categories']).lower().split()
        )

        # cek apakah ada kata kategori yang sama
        rel = 1 if len(kategori_asli & kategori_rekom) > 0 else 0

        relevansi.append(rel)

    # DCG
    dcg_val = sum([
        rel / np.log2(i + 2)
        for i, rel in enumerate(relevansi)
    ])

    # IDCG
    ideal = sorted(relevansi, reverse=True)

    idcg_val = sum([
        rel / np.log2(i + 2)
        for i, rel in enumerate(ideal)
    ])

    if idcg_val == 0:
        return 0

    return dcg_val / idcg_val
```

```python
def evaluasi_sistem(data, cosine_sim, k=3):

    start_time = time.time()   # mulai hitung waktu

    total_precision = 0
    total_ndcg = 0

    for idx in range(len(data)):

        total_precision += precision_at_k(
            idx,
            cosine_sim,
            data,
            k
        )

        total_ndcg += ndcg_at_k(
            idx,
            cosine_sim,
            data,
            k
        )

    avg_precision = total_precision / len(data)
    avg_ndcg = total_ndcg / len(data)

    end_time = time.time()   # selesai hitung waktu
    running_time = end_time - start_time

    print("Precision@{} :".format(k), round(avg_precision, 3))
    print("NDCG@{}      :".format(k), round(avg_ndcg, 3))
    print("Running Time :", round(running_time, 4), "detik")

    return avg_precision, avg_ndcg, running_time
```

```python
rekomendasi("harry potter")

evaluasi_sistem(data, cosine_sim, k=3)
evaluasi_sistem(data, cosine_sim, k=5)
evaluasi_sistem(data, cosine_sim, k=10)
```
